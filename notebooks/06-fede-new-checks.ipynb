```julia
using LinearAlgebra
```

```julia
bprob(p1) = p1, 1-p1

p1, p2 = bprob(0.35)
q1, q2 = bprob(0.01)
```

```julia
k0 = [sqrt(q1/p1) 0.0; 0.0 1]

k1 = [0.0 0.0; sqrt(1-q1/p1) 0]

display(k0)
display(k1)
```

```julia
k0' * k0+k1'*k1
```

```julia
P=[p1 0.0; 0.0 p2]
Q = [q1 0.0; 0.0 q2]
```

```julia
gamma_action(x::Matrix{Float64}) = k0 * x * k0' + k1 * x * k1'

Q_new = gamma_action(P)
isapprox(Q, Q_new)
```
